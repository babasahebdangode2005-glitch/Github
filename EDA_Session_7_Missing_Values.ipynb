**KNNImputer**

- KNN: K-nearest neighbours

- where K is number

- instaed of taking avg of all the values data

    - first will find the neighbours who are close with missing data

    - by using Euclidian distance calculation will idenify the neighbour rows

    - suppose k=2 will identify the two closest rows

    - those closest rows values will take it, will calculate the mean

    - finally will fill with the mean

```python
salary=[50000,np.nan,70000,90000]
age=[25,28,30,40]
df=pd.DataFrame(zip(salary,age),
                columns=['Salary','Age'])
df
```

- sklearn

    - impute

        - KKNImputer


```python
from sklearn.impute import KNNImputer
imputer = KNNImputer(n_neighbors=2)

pd.DataFrame(imputer.fit_transform(df))
```
